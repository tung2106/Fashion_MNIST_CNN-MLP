```python
from pathlib import Path  # Xử lý đường dẫn.
import sys  # Lấy đúng Python đang chạy notebook.
import subprocess  # Gọi chương trình Python riêng.

# Sửa duy nhất đường dẫn này cho đúng nơi bạn đặt dự án.
PROJECT = Path("/content/Fashion_MNIST_CNN-MLP")
STAGE = PROJECT / "03_thuc_nghiem"
if not (STAGE / "main.py").is_file():  # Hỗ trợ bản cũ trên GitHub.
    STAGE = PROJECT / "03_Thuc_Nghiem_Cai_Tien"
assert (STAGE / "main.py").exists(), "Cần đặt mã giai đoạn 3 vào PROJECT trước."
```

## 1. Chỉ xem dữ liệu
