# Chạy các module của dự án

Kiểm tra dữ liệu có thể chạy local. Các ô train/smoke chỉ chạy Colab GPU. Nếu dùng Colab, giải nén gói source.zip vào PROJECT trước.


```python
import subprocess
import sys
from pathlib import Path

PROJECT = next(
    (
        p
        for p in [Path.cwd(), *Path.cwd().parents]
        if (p / "training/huan_luyen.py").exists()
    ),
    Path("/content/Fashion_MNIST_CNN-MLP"),
)
assert (PROJECT / "training/huan_luyen.py").exists(), "Chỉnh PROJECT tới gốc dự án."
sys.path.insert(0, str(PROJECT))
```

## 1. Chỉ xem dữ liệu

Xem split và shape của một batch.


```python
subprocess.run(
    [sys.executable, "-m", "training.huan_luyen", "--inspect"], cwd=PROJECT, check=True
)
```

## 2. Chạy thử nhỏ

Kiểm tra hai mô hình trên 320 ảnh train và 80 ảnh validation.


```python
subprocess.run(
    [sys.executable, "-m", "training.huan_luyen", "--config", "configs/chay_thu.json"],
    cwd=PROJECT,
    check=True,
)
```
